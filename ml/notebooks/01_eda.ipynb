# 01. Exploratory Data Analysis (EDA)
This notebook performs exploratory data analysis on raw extracted quiz attempts from the Video Intelligence Platform.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

df = pd.read_csv('../data/raw/quiz_attempts.csv')
print(f"Total Quiz Attempts: {len(df)}")
print(f"Total Unique Users: {df['user_id'].nunique()}")
df.head()
```

```python
df.describe()
```

```python
attempts_per_user = df.groupby('user_id').size()
print(attempts_per_user.describe())

plt.figure(figsize=(8, 4))
sns.histplot(attempts_per_user, bins=12, discrete=True, color='#6366f1')
plt.title('Attempts Per User Distribution')
plt.xlabel('Number of Attempts')
plt.ylabel('Users')
plt.show()
```

```python
plt.figure(figsize=(8, 4))
sns.histplot(df['percentage'], bins=15, kde=True, color='#14b8a6')
plt.title('Quiz Score Percentage Distribution')
plt.xlabel('Percentage (%)')
plt.ylabel('Count')
plt.show()
```
